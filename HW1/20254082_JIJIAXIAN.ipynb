```python
import pandas as pd
df = pd.read_csv('movies_metadata.csv', low_memory=False)
df = df[['title', 'overview']]
df = df.dropna(subset=['overview'])

from sklearn.feature_extraction.text import TfidfVectorizer
tfidf = TfidfVectorizer(stop_words='english')
tfidf_matrix = tfidf.fit_transform(df['overview'])

from sklearn.metrics.pairwise import cosine_similarity
```

```python
indices = pd.Series(df.index, index=df['title']).drop_duplicates()
cosine_sim = cosine_similarity(tfidf_matrix, tfidf_matrix)

def get_recommendations(title, cosine_sim=cosine_sim):
    idx = indices[title]
    sim_scores = list(enumerate(cosine_sim[idx]))
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)
    sim_scores = sim_scores[1:11]

    movie_indices = [i[0] for i in sim_scores]
    return df['title'].iloc[movie_indices]

# Ask the user for input and store it in the variable 'user_movie'
user_movie = input("Please enter a movie title to find similar movies: ")

# Safety check: see if the movie exists in our 'indices' dictionary
if user_movie in indices:
    print(f"\nHere are the top 10 movies most similar to '{user_movie}':")
    print("-" * 40)
    print(get_recommendations(user_movie))
else:
    print(f"\nSorry, the movie '{user_movie}' is not in our database. Please check your spelling and capitalization!")


```
